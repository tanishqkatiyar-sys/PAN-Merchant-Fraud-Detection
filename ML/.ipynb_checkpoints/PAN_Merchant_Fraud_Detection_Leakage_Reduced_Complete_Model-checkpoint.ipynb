# PAN Merchant Fraud Detection — EDA & Leakage-Aware Modeling

This notebook explores a synthetic merchant dataset and compares behavioral fraud-detection models.

**Leakage-aware design:** graph edges and graph embeddings are retained only for exploratory network analysis. They are not used as predictors because the supplied edge list was constructed using fraud labels. Shared-identifier counts, chargeback rate and refund ratio are also excluded until their provenance and availability at prediction time are verified.

The primary model uses a conservative whitelist of merchant profile and transaction-volume/value features. This reduces known leakage risks but does not prove that every retained feature is leakage-free; inspect the data-generation process before making real-world claims.

Keep `merchant_synthetic_100k.csv` and `merchant_edges.csv` in the notebook's working directory, or update the paths in the data-loading cell.

## 2. Load the Data

We load both CSV files generated by the data-generation notebook/pipeline.

```python
from pathlib import Path

def first_existing(candidates):
    for candidate in candidates:
        path = Path(candidate)
        if path.exists():
            return str(path)
    raise FileNotFoundError(
        "Could not find any of these files: " + ", ".join(candidates) +
        ". Put the CSV files beside the notebook or update the candidate paths."
    )

merchant_path = first_existing([
    "merchant_synthetic_100k.csv",
    "merchant_synthetic_100k(1).csv",
    "/mnt/data/merchant_synthetic_100k.csv",
    "/mnt/data/merchant_synthetic_100k(1).csv",
])
edges_path = first_existing([
    "merchant_edges.csv",
    "merchant_edges(1).csv",
    "/mnt/data/merchant_edges.csv",
    "/mnt/data/merchant_edges(1).csv",
])

merchants = pd.read_csv(merchant_path)
edges = pd.read_csv(edges_path)

print("Merchant file:", merchant_path, "| shape:", merchants.shape)
print("Edge file:", edges_path, "| shape:", edges.shape)
merchants.head()
```

## 2.1 Diagnose the stored embeddings

This checks whether the `emb_0`–`emb_15` values in the loaded CSV are actually zero, constant, missing, or varied. A zero vector from graph-SVD can occur for an isolated node, but stored embeddings may have been generated separately; their provenance cannot be inferred from column names alone. These columns remain excluded from predictive training because graph data is label-informed / provenance is unverified.


```python
embedding_cols = [c for c in merchants.columns if c.startswith('emb_')]
if not embedding_cols:
    print('No stored embedding columns found in this CSV.')
else:
    E = merchants[embedding_cols].apply(pd.to_numeric, errors='coerce')
    all_zero_mask = np.isclose(E.fillna(0).to_numpy(), 0).all(axis=1)
    print(f'Embedding columns: {len(embedding_cols)}')
    print(f'Rows with all-zero embeddings: {all_zero_mask.sum():,} / {len(E):,} ({all_zero_mask.mean():.2%})')
    print(f'Rows with any missing embedding value: {E.isna().any(axis=1).sum():,}')
    print('Non-zero values per dimension:')
    display(E.ne(0).sum().to_frame('nonzero_count').T)
    print('Unique values per dimension (first 16):')
    display(E.nunique(dropna=True).to_frame('unique_count').T)
    print('First five stored vectors:')
    display(E.head())
    if all_zero_mask.all():
        print('Diagnosis: all stored vectors are zero. Check the embedding-generation step, node order / merchant_id join, and whether the source graph has edges for these nodes.')
    elif all_zero_mask.any():
        print('Diagnosis: only some rows are all-zero. These may be isolated/unrepresented nodes or failed/missing embeddings; inspect them against graph degree and generation logs.')
    else:
        print('Diagnosis: the loaded CSV does not contain all-zero embedding rows. If another displayed table shows zeros, verify it was generated from this same CSV and was not rounded/overwritten.')
```

### 6.1 Graph features from the edge list

Graph degree and weighted degree are calculated for exploratory network analysis only. The supplied graph was generated using fraud labels, so these fields are not used as predictors in the primary model.


```python
# Calculate graph metrics for exploratory analysis only.
# Do not use these features in the primary fraud-prediction model.
required_edge_cols = {"merchant_A", "merchant_B"}
if not required_edge_cols.issubset(edges.columns):
    raise ValueError(f"Edge file must contain columns: {sorted(required_edge_cols)}")

edge_degree = pd.concat(
    [edges["merchant_A"].astype(str), edges["merchant_B"].astype(str)]
).value_counts()
merchants["edge_degree"] = merchants["merchant_id"].astype(str).map(edge_degree).fillna(0).astype(int)

if "weight" in edges.columns:
    edge_weight = pd.concat([
        edges.assign(_merchant=edges["merchant_A"].astype(str)).groupby("_merchant")["weight"].sum(),
        edges.assign(_merchant=edges["merchant_B"].astype(str)).groupby("_merchant")["weight"].sum(),
    ]).groupby(level=0).sum()
    merchants["edge_weight_sum"] = merchants["merchant_id"].astype(str).map(edge_weight).fillna(0)
else:
    merchants["edge_weight_sum"] = merchants["edge_degree"].astype(float)

print("Graph metrics created for EDA only.")
print("Merchants with graph connections:", int((merchants["edge_degree"] > 0).sum()))
```

### ⚠️ Important data-leakage note

The supplied edge list is understood to have been generated using fraud labels and therefore may encode the target directly. Graph connectivity, edge-derived measures and embeddings are used only for exploratory network analysis and are excluded from the primary predictive model.

The edge-degree/label cross-tabulation below is a diagnostic, not a model result. No predictive model using these graph features is trained or reported.


```python
leak_corr = merchants["edge_degree"].corr(merchants["is_fraud"])
print(f"Correlation between edge_degree and is_fraud (EDA diagnostic only): {leak_corr:.4f}")
pd.crosstab(merchants["edge_degree"] > 0, merchants["is_fraud"],
            rownames=["Has graph connection"], colnames=["Fraud label"])
```

### 6.2 Generate Merchant Embeddings (Graph-based)

Embeddings are generated here only to demonstrate exploratory graph representation and similarity analysis. Because the supplied edge list was constructed using fraud labels, these embeddings are potentially target-leaking and are **not included in the primary predictive model**.


```python
import networkx as nx
from sklearn.decomposition import TruncatedSVD

EMBED_DIM = 16 


G = nx.Graph()
G.add_nodes_from(merchants["merchant_id"])
G.add_weighted_edges_from(edges[["merchant_A", "merchant_B", "weight"]].values)


merchant_order = merchants["merchant_id"].tolist()


adjacency_matrix = nx.to_scipy_sparse_array(G, nodelist=merchant_order, weight="weight")
print("Adjacency matrix shape:", adjacency_matrix.shape)


svd = TruncatedSVD(n_components=EMBED_DIM, random_state=42)
embeddings = svd.fit_transform(adjacency_matrix)
print("Embeddings shape:", embeddings.shape)


embedding_df = pd.DataFrame(embeddings, columns=[f"emb_{i}" for i in range(EMBED_DIM)])
embedding_df["merchant_id"] = merchant_order

merchants = merchants.merge(embedding_df, on="merchant_id", how="left")
merchants.filter(like="emb_").head()
```

Embeddings below are exploratory only because they are derived from the label-informed edge list. They are not predictors in the primary model. Isolated nodes may have zero vectors; cosine similarity involving zero vectors should be interpreted cautiously.


```python
embedding_cols = [c for c in merchants.columns if c.startswith("emb_")]
n_isolated = (np.isclose(merchants[embedding_cols].to_numpy(), 0).all(axis=1)).sum()
print(f"Merchants with an all-zero embedding: {n_isolated:,} out of {len(merchants):,}")
# Save only as an optional exploratory artifact; this does not replace the source CSV.
merchants.to_csv("merchant_synthetic_100k_with_embeddings_exploratory.csv", index=False)
print("Saved exploratory embedding dataset.")
```

### 6.3 Cosine Similarity — Finding Similar Merchants

As described in the project README, the system uses **cosine similarity** on
these embeddings to find merchants that "look alike" in connection-pattern terms —
this is exactly how a fraud ring can be expanded: start from one known fraud
merchant and find others with a very similar embedding.

**Cosine similarity** measures the angle between two vectors:
- `1.0` → pointing in exactly the same direction (very similar)
- `0.0` → unrelated
- `-1.0` → pointing in opposite directions

We write one small, reusable function that, given a merchant ID, returns the
merchants most similar to it.

```python
from sklearn.metrics.pairwise import cosine_similarity

embedding_cols = [c for c in merchants.columns if c.startswith("emb_")]
embedding_matrix = merchants[embedding_cols].values
merchant_id_list = merchants["merchant_id"].tolist()
id_to_row = {mid: i for i, mid in enumerate(merchant_id_list)}


def find_similar_merchants(merchant_id, top_n=5):
    """Return the top_n merchants whose embedding is most similar to merchant_id's."""
    row_idx = id_to_row[merchant_id]
    target_vector = embedding_matrix[row_idx].reshape(1, -1)


    similarities = cosine_similarity(target_vector, embedding_matrix)[0]

    result = merchants[["merchant_id", "is_fraud"]].copy()
    result["similarity"] = similarities
    result = result[result["merchant_id"] != merchant_id] 
    return result.sort_values("similarity", ascending=False).head(top_n)
```

**Demo:** select a connected merchant for an exploratory similarity example. This is not an independent fraud-prediction evaluation because the graph is label-informed.


```python
connected = merchants.loc[merchants["edge_degree"] > 0, "merchant_id"]
if len(connected) == 0:
    print("No connected merchants are present in the edge list; skipping similarity demo.")
else:
    example_fraud_merchant = connected.iloc[0]
    print(f"Exploratory similarity for merchant: {example_fraud_merchant}")
    display(find_similar_merchants(example_fraud_merchant, top_n=5))
```

**Validation:** on average, do merchants *within the same fraud ring* have higher
cosine similarity to each other than two *random, unrelated* merchants? If our
embeddings are working, the answer should be a clear yes.

```python
import random
random.seed(42)

if len(edges) == 0:
    print("No edges available; skipping pair-similarity comparison.")
else:
    n_pairs = min(300, len(edges), len(merchant_id_list) // 2)
    same_ring_pairs = edges[["merchant_A", "merchant_B"]].sample(n_pairs, random_state=42).values
    random_ids = random.sample(merchant_id_list, 2 * n_pairs)
    random_pairs = list(zip(random_ids[:n_pairs], random_ids[n_pairs:]))

    def pair_similarity(pairs):
        sims = []
        for a, b in pairs:
            va = embedding_matrix[id_to_row[a]].reshape(1, -1)
            vb = embedding_matrix[id_to_row[b]].reshape(1, -1)
            # A zero vector has undefined cosine similarity; record NaN.
            if np.linalg.norm(va) == 0 or np.linalg.norm(vb) == 0:
                sims.append(np.nan)
            else:
                sims.append(cosine_similarity(va, vb)[0, 0])
        return sims

    same_ring_sims = pair_similarity(same_ring_pairs)
    random_sims = pair_similarity(random_pairs)
    sim_df = pd.DataFrame({
        "similarity": same_ring_sims + random_sims,
        "pair_type": ["Connected edge pair"] * len(same_ring_sims) + ["Random pair"] * len(random_sims),
    })
    print(sim_df.groupby("pair_type")["similarity"].mean())
    plt.figure(figsize=(6, 4))
    sns.boxplot(data=sim_df, x="pair_type", y="similarity")
    plt.title("Exploratory Cosine Similarity: Edge Pairs vs Random Pairs")
    plt.ylabel("Cosine similarity")
    plt.xlabel("")
    plt.tight_layout()
    plt.show()
```

## 7. Build the Modeling Table

The primary model uses a conservative, explicitly listed set of profile and transaction-volume/value features. Identifiers, graph features, embeddings, shared-identifier counts, chargeback rate and refund ratio are excluded because their prediction-time availability or label-independent provenance has not been established.


```python
# Conservative feature whitelist. Confirm each feature is available at the intended
# prediction time and was not generated conditionally on the fraud label.
target_col = "is_fraud"
numeric_features = [
    "is_kyc_verified", "total_txns_90d", "total_txns_30d", "total_txns_7d",
    "avg_txn_value", "avg_txn_value_30d", "median_txn_value_90d",
    "std_txn_value_90d", "min_txn_value_30d", "max_txn_value_30d",
    "pct_high_value_txns", "high_value_txns_90d", "high_value_txns_30d",
    "merchant_age_days", "days_since_registration",
]
categorical_features = ["merchant_category", "business_city", "merchant_tier"]

required = [target_col] + numeric_features + categorical_features
missing_cols = [c for c in required if c not in merchants.columns]
if missing_cols:
    raise ValueError(f"Missing required columns: {missing_cols}")

model_df = merchants[required].copy()
model_df[target_col] = pd.to_numeric(model_df[target_col], errors="raise").astype(int)
if not set(model_df[target_col].dropna().unique()).issubset({0, 1}):
    raise ValueError("Target 'is_fraud' must contain only binary values 0 and 1.")

for col in numeric_features:
    model_df[col] = pd.to_numeric(model_df[col], errors="coerce")
for col in categorical_features:
    model_df[col] = model_df[col].fillna("Unknown").astype(str)

y = model_df[target_col]
X = model_df[numeric_features + categorical_features].copy()
print("Primary model uses the conservative feature whitelist.")
print("Excluded: graph/embedding, shared-identifier, chargeback and refund features.")
print("X shape:", X.shape)
```

## 8. Train / Test Split

A stratified 80/20 split preserves the target proportion in each partition. Preprocessing is fitted on training data only and then applied to the test data.


```python
idx_train, idx_test = train_test_split(
    model_df.index, test_size=0.2, random_state=42, stratify=y
)
X_train_raw = X.loc[idx_train].copy()
X_test_raw = X.loc[idx_test].copy()
y_train = y.loc[idx_train].copy()
y_test = y.loc[idx_test].copy()

# Fit numeric imputation values on training rows only.
train_medians = X_train_raw[numeric_features].median()
X_train_raw[numeric_features] = X_train_raw[numeric_features].fillna(train_medians)
X_test_raw[numeric_features] = X_test_raw[numeric_features].fillna(train_medians).fillna(0)

# One-hot encode separately, then align test columns to the training schema.
Xb_train = pd.get_dummies(X_train_raw, columns=categorical_features, dummy_na=False)
Xb_test = pd.get_dummies(X_test_raw, columns=categorical_features, dummy_na=False)
Xb_test = Xb_test.reindex(columns=Xb_train.columns, fill_value=0)

print("Train size:", len(idx_train), "| Test size:", len(idx_test))
print("Fraud rate — train:", round(y_train.mean(), 3), "| test:", round(y_test.mean(), 3))
print("Encoded train/test shapes:", Xb_train.shape, Xb_test.shape)
```

## 9. Model Comparison

Models are compared using the same stratified holdout and conservative feature whitelist. Graph-derived features, embeddings and unverified post-outcome or shared-identifier features are excluded. This synthetic-data experiment does not establish production performance; a time-based split and independent validation are needed for deployment-oriented claims.


```python
def evaluate(name, model, X_train, X_test, y_train, y_test, scale=False):
    """Fit on training data and evaluate once on the held-out test data."""
    if scale:
        scaler = StandardScaler()
        X_train_fit = scaler.fit_transform(X_train)
        X_test_fit = scaler.transform(X_test)
    else:
        X_train_fit, X_test_fit = X_train, X_test

    model.fit(X_train_fit, y_train)
    proba = model.predict_proba(X_test_fit)[:, 1]
    pred = model.predict(X_test_fit)

    auc = roc_auc_score(y_test, proba)
    f1 = f1_score(y_test, pred, zero_division=0)
    print(f"{name:28s}  ROC-AUC = {auc:.4f}   F1 = {f1:.4f}")
    return {"name": name, "model": model, "auc": auc, "f1": f1,
            "proba": proba, "pred": pred}
```

```python
print("--- Leakage-reduced behavioral feature comparison ---")
results_behavioral = [
    evaluate("Logistic Regression", LogisticRegression(max_iter=500, class_weight="balanced"),
             Xb_train, Xb_test, y_train, y_test, scale=True),
    evaluate("Random Forest", RandomForestClassifier(
        n_estimators=200, max_depth=12, class_weight="balanced",
        random_state=42, n_jobs=-1), Xb_train, Xb_test, y_train, y_test),
    evaluate("HistGradientBoosting", HistGradientBoostingClassifier(
        max_iter=200, random_state=42, class_weight="balanced"),
             Xb_train, Xb_test, y_train, y_test),
]
```

```python
# Deliberately no "Full (+ graph)" model is trained here:
# its input graph is label-derived, so such scores would be misleading.
print("Graph and embedding features are excluded from predictive model training.")
```

```python
comparison = pd.DataFrame([
    {"Feature set": "Leakage-reduced behavioral", **{k: r[k] for k in ("name", "auc", "f1")}}
    for r in results_behavioral
]).rename(columns={"name": "Model", "auc": "ROC-AUC", "f1": "F1-score"})

comparison.sort_values("ROC-AUC", ascending=False)
```

**Interpretation:** Only the leakage-reduced behavioral feature set is evaluated as a predictive model. The graph-based features and embeddings are excluded because the supplied edge list is label-derived. Results should be reported as performance on this synthetic holdout, not as guaranteed real-world performance.


## 10. Final Model Evaluation

For continuity, the Random Forest is shown as a baseline final model. Its selection here is illustrative, not a claim that it is universally optimal. For rigorous selection, use cross-validation or a separate validation set, then evaluate once on the untouched test set.


```python
final = next(r for r in results_behavioral if r["name"] == "Random Forest")
print(f"Final model: {final['name']}")
print(f"ROC-AUC: {final['auc']:.4f}")
print(f"F1-score: {final['f1']:.4f}")

print("\nFull classification report:")
print(classification_report(y_test, final["pred"], target_names=["Legit", "Fraud"]))
```

## 11. Conclusion

- The dataset is synthetic; results describe this generated dataset and should not be generalized to real merchants without external validation.
- The supplied graph is label-informed, so graph connectivity, graph-derived features and embeddings are excluded from predictive training.
- Shared-identifier counts, chargeback rate and refund ratio are excluded until their generation process and availability at prediction time are verified.
- The primary model uses a conservative whitelist of merchant profile and transaction-volume/value features.
- Numeric imputation values and categorical encodings are fitted using training data only.
- Metrics must be regenerated by running the notebook from a clean kernel. They represent a stratified holdout result, not a production guarantee.
- Confirm the provenance and prediction-time availability of every retained feature. If any retained feature uses the label or future information, remove or regenerate it.
- For deployment-oriented evaluation, use a time-based split, a documented prediction cutoff, leakage-safe feature generation and independent data.

It does not use `is_fraud`, identifiers, graph edges, embeddings, shared-identifier counts, chargeback rate, or refund ratio as predictors. Confirm feature provenance and prediction-time availability before treating this as deployment-ready.


```python
# Train the final model on the training partition only; test data remains held out.
final_model = RandomForestClassifier(n_estimators=300, max_depth=12,
    class_weight='balanced', random_state=42, n_jobs=-1)
final_model.fit(Xb_train, y_train)

def predict_new_merchants(new_data, threshold=0.50):
    """Return fraud probability and threshold-based prediction for new merchant rows."""
    if not 0 < threshold < 1:
        raise ValueError('threshold must be between 0 and 1')
    missing = [c for c in numeric_features + categorical_features if c not in new_data.columns]
    if missing:
        raise ValueError(f'Missing required prediction fields: {missing}')
    z = new_data[numeric_features + categorical_features].copy()
    for col in numeric_features:
        z[col] = pd.to_numeric(z[col], errors='coerce').fillna(train_medians[col])
        if z[col].isna().all():
            z[col] = z[col].fillna(0)
    for col in categorical_features:
        z[col] = z[col].fillna('Unknown').astype(str)
    z = pd.get_dummies(z, columns=categorical_features, dummy_na=False)
    z = z.reindex(columns=Xb_train.columns, fill_value=0)
    prob = final_model.predict_proba(z)[:, 1]
    return pd.DataFrame({'fraud_probability': prob,
        'predicted_label': (prob >= threshold).astype(int)}, index=new_data.index)

# Example: predict the first five held-out rows without accessing their labels.
example_input = merchants.loc[idx_test, numeric_features + categorical_features].head(5).copy()
example_predictions = predict_new_merchants(example_input, threshold=0.50)
display(example_predictions)

# Save model plus preprocessing schema for later reuse.
import joblib
model_bundle = {'model': final_model, 'numeric_features': numeric_features,
    'categorical_features': categorical_features, 'train_medians': train_medians,
    'encoded_columns': list(Xb_train.columns), 'threshold': 0.50,
    'excluded_features': ['is_fraud', 'merchant_id', 'graph features', 'embeddings',
        'shared-identifier counts', 'chargeback_rate', 'refund_ratio']}
joblib.dump(model_bundle, 'pan_merchant_fraud_model.joblib')
print('Saved: pan_merchant_fraud_model.joblib')
```
