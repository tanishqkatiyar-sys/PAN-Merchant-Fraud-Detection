# PAN Merchant Fraud Detection — EDA & Leakage-Aware Modeling

This notebook explores a synthetic merchant dataset and compares behavioral fraud-detection models.

**Leakage-aware design:** graph edges and graph embeddings are retained only for exploratory network analysis. They are not used as predictors because the supplied edge list was constructed using fraud labels. Post-outcome / potentially target-derived fields (chargeback and refund rates) and shared-identifier counts are also excluded from the primary model until their generation time and label independence can be verified.

The model uses a conservative whitelist of merchant profile and transaction-volume/value features. Even these must be interpreted in the context of the synthetic data-generation process; correlation alone does not prove leakage-free provenance.

Place `merchant_synthetic_100k(1).csv` and `merchant_edges(1).csv` in the same folder as this notebook.

## 4. Missing Value Check & Handling

We check every column for nulls. For a real-world dataset you would typically see
missing values — here we handle them generically so the same code works even if
you plug in noisier data later:
- **Numeric columns** → filled with the **median** (robust to outliers)
- **Categorical columns** → filled with the **mode** (most frequent value)

```python
missing = merchants.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
print("Columns with missing values:")
print(missing if len(missing) else "None found ✅")
```

```python
num_cols = merchants.select_dtypes(include=["int64", "float64"]).columns
cat_cols = merchants.select_dtypes(include=["object"]).columns

for c in num_cols:
    if merchants[c].isnull().any():
        merchants[c] = merchants[c].fillna(merchants[c].median())

for c in cat_cols:
    if merchants[c].isnull().any():
        merchants[c] = merchants[c].fillna(merchants[c].mode()[0])

print("Total missing values after cleaning:", merchants.isnull().sum().sum())
```

### 6.1 Graph features from the edge list

Graph degree and weighted degree are calculated for exploratory network analysis only. The supplied graph was generated using fraud labels, so these fields are not used as predictors in the primary model.


```python
### ⚠️ Graph-label leakage warning

The supplied synthetic edge list was constructed using the fraud labels. In this data, graph connectivity and graph-derived measures can directly reveal the target. The same applies to embeddings generated from this graph: the combined all-zero/non-zero embedding pattern can separate connected and isolated merchants.

Therefore, `edge_degree`, `edge_weight_sum`, and all `emb_*` columns are **exploratory only** and are excluded from the primary predictive model. A graph-model score based on these data would be contaminated and must not be presented as realistic performance.
```

### ⚠️ Important data-leakage note

In this synthetic dataset, `merchant_edges.csv` was generated by first choosing which
merchants are fraudulent and *then* connecting those fraudulent merchants together into
rings. That means **every merchant with `edge_degree > 0` is fraudulent, with no
exceptions** — this single feature gives a model ~100% accuracy, which would be
unrealistic for a real fraud system (in the real world, you wouldn't know who's fraudulent
before building the graph!).

We confirm this below, then deliberately train our **final model without the graph/edge
features**, so the reported performance reflects what a model could realistically
achieve from transaction behavior alone. We still report the "with graph features"
result afterwards for comparison — it demonstrates the extra lift a network-analysis
approach *would* give if the shared-identifier signal were partially known in advance.

```python
leak_corr = merchants["edge_degree"].corr(merchants["is_fraud"])
print(f"Correlation between edge_degree and is_fraud: {leak_corr:.4f}")
pd.crosstab(merchants["edge_degree"] > 0, merchants["is_fraud"])
```

### 6.2 Generate Merchant Embeddings (Graph-based)

Embeddings are generated here only to demonstrate exploratory graph representation and similarity analysis. Because the supplied edge list was constructed using fraud labels, these embeddings are potentially target-leaking and are **not included in the primary predictive model**.


```python
import networkx as nx
from sklearn.decomposition import TruncatedSVD

EMBED_DIM = 16 


G = nx.Graph()
G.add_nodes_from(merchants["merchant_id"])
G.add_weighted_edges_from(edges[["merchant_A", "merchant_B", "weight"]].values)


merchant_order = merchants["merchant_id"].tolist()


adjacency_matrix = nx.to_scipy_sparse_array(G, nodelist=merchant_order, weight="weight")
print("Adjacency matrix shape:", adjacency_matrix.shape)


svd = TruncatedSVD(n_components=EMBED_DIM, random_state=42)
embeddings = svd.fit_transform(adjacency_matrix)
print("Embeddings shape:", embeddings.shape)


embedding_df = pd.DataFrame(embeddings, columns=[f"emb_{i}" for i in range(EMBED_DIM)])
embedding_df["merchant_id"] = merchant_order

merchants = merchants.merge(embedding_df, on="merchant_id", how="left")
merchants.filter(like="emb_").head()
```

Merchants with no shared PAN/device/IP connections get an all-zero embedding —
that's expected, since there's nothing to compress for an isolated node. We can
confirm that below, then save the enriched dataset.

```python
n_isolated = (merchants.filter(like="emb_").sum(axis=1) == 0).sum()
print(f"Merchants with an all-zero embedding (no connections): {n_isolated:,} out of {len(merchants):,}")

merchants.to_csv("merchant_synthetic_100k_with_embeddings.csv", index=False)
print("Saved: merchant_synthetic_100k_with_embeddings.csv")
```

### 6.3 Cosine Similarity — Finding Similar Merchants

As described in the project README, the system uses **cosine similarity** on
these embeddings to find merchants that "look alike" in connection-pattern terms —
this is exactly how a fraud ring can be expanded: start from one known fraud
merchant and find others with a very similar embedding.

**Cosine similarity** measures the angle between two vectors:
- `1.0` → pointing in exactly the same direction (very similar)
- `0.0` → unrelated
- `-1.0` → pointing in opposite directions

We write one small, reusable function that, given a merchant ID, returns the
merchants most similar to it.

```python
from sklearn.metrics.pairwise import cosine_similarity

embedding_cols = [c for c in merchants.columns if c.startswith("emb_")]
embedding_matrix = merchants[embedding_cols].values
merchant_id_list = merchants["merchant_id"].tolist()
id_to_row = {mid: i for i, mid in enumerate(merchant_id_list)}


def find_similar_merchants(merchant_id, top_n=5):
    """Return the top_n merchants whose embedding is most similar to merchant_id's."""
    row_idx = id_to_row[merchant_id]
    target_vector = embedding_matrix[row_idx].reshape(1, -1)


    similarities = cosine_similarity(target_vector, embedding_matrix)[0]

    result = merchants[["merchant_id", "is_fraud"]].copy()
    result["similarity"] = similarities
    result = result[result["merchant_id"] != merchant_id] 
    return result.sort_values("similarity", ascending=False).head(top_n)
```

**Demo:** pick a merchant that we know is part of a fraud ring (`edge_degree > 0`)
and see whether its most similar merchants are also flagged as fraud.

```python
example_fraud_merchant = merchants.loc[merchants["edge_degree"] > 0, "merchant_id"].iloc[0]
print(f"Looking for merchants similar to: {example_fraud_merchant}")

find_similar_merchants(example_fraud_merchant, top_n=5)
```

**Validation:** on average, do merchants *within the same fraud ring* have higher
cosine similarity to each other than two *random, unrelated* merchants? If our
embeddings are working, the answer should be a clear yes.

```python
import random
random.seed(42)


same_ring_pairs = edges[["merchant_A", "merchant_B"]].sample(300, random_state=42).values


random_ids = random.sample(merchant_id_list, 600)
random_pairs = list(zip(random_ids[:300], random_ids[300:]))


def pair_similarity(pairs):
    sims = []
    for a, b in pairs:
        va = embedding_matrix[id_to_row[a]].reshape(1, -1)
        vb = embedding_matrix[id_to_row[b]].reshape(1, -1)
        sims.append(cosine_similarity(va, vb)[0, 0])
    return sims

same_ring_sims = pair_similarity(same_ring_pairs)
random_sims = pair_similarity(random_pairs)

sim_df = pd.DataFrame({
    "similarity": same_ring_sims + random_sims,
    "pair_type": ["Same fraud ring"] * len(same_ring_sims) + ["Random pair"] * len(random_sims),
})

print(sim_df.groupby("pair_type")["similarity"].mean())

plt.figure(figsize=(6, 4))
sns.boxplot(data=sim_df, x="pair_type", y="similarity")
plt.title("Cosine Similarity: Same Fraud Ring vs Random Pairs")
plt.ylabel("Cosine similarity")
plt.xlabel("")
plt.tight_layout()
plt.show()
```

## 7. Build the Modeling Table

We drop columns that are not useful as direct model inputs:
- `merchant_id` — just an identifier
- raw hash columns (`pan_hash`, `device_id_hash`, etc.) — already summarized by the
  `shared_*_count` features
- original categorical text columns — replaced by their `_enc` encoded versions
- `merchant_registration_date` — replaced by `days_since_registration`

We then prepare **two** feature sets so we can honestly compare performance:
- `X_behavioral` — transaction/profile features only (leakage-reduced)
- `X_full` — behavioral + graph/embedding features (includes the leakage discussed above)

```python
# Conservative feature whitelist. Exclude:
# - identifiers and raw hashes (not direct predictors)
# - target label
# - graph features and embeddings (edge list is label-derived)
# - shared-identifier counts (provenance/availability must be verified)
# - chargeback_rate and refund_ratio (may include post-outcome information)
# - original date and text columns (categoricals are encoded after splitting)
target_col = "is_fraud"
numeric_features = [
    "is_kyc_verified",
    "total_txns_90d",
    "total_txns_30d",
    "total_txns_7d",
    "avg_txn_value",
    "avg_txn_value_30d",
    "median_txn_value_90d",
    "std_txn_value_90d",
    "min_txn_value_30d",
    "max_txn_value_30d",
    "pct_high_value_txns",
    "high_value_txns_90d",
    "high_value_txns_30d",
    "merchant_age_days",
    "days_since_registration",
]
categorical_features = ["merchant_category", "business_city", "merchant_tier"]

required = [target_col] + numeric_features + categorical_features
missing_cols = [c for c in required if c not in merchants.columns]
if missing_cols:
    raise ValueError(f"Missing required columns: {missing_cols}")

model_df = merchants[required].copy()
model_df[target_col] = pd.to_numeric(model_df[target_col], errors="raise").astype(int)
model_df[numeric_features] = model_df[numeric_features].apply(pd.to_numeric, errors="coerce")
for col in numeric_features:
    model_df[col] = model_df[col].fillna(model_df[col].median())
for col in categorical_features:
    model_df[col] = model_df[col].fillna("Unknown").astype(str)

y = model_df[target_col]
X = model_df[numeric_features + categorical_features].copy()
print("Primary model uses only the documented conservative feature whitelist.")
print("Excluded graph, embedding, shared-identifier, chargeback and refund features.")
print("X shape:", X.shape)
```

## 8. Train / Test Split

We use a stratified 80/20 split so the fraud rate is preserved in both sets, and
reuse the **same row indices** for both feature sets so comparisons are fair.

```python
idx_train, idx_test = train_test_split(
    model_df.index, test_size=0.2, random_state=42, stratify=y
)

X_train_raw = X.loc[idx_train].copy()
X_test_raw = X.loc[idx_test].copy()
y_train = y.loc[idx_train].copy()
y_test = y.loc[idx_test].copy()

# Fit categorical representation on training data only, then align test columns.
Xb_train = pd.get_dummies(X_train_raw, columns=categorical_features, dummy_na=False)
Xb_test = pd.get_dummies(X_test_raw, columns=categorical_features, dummy_na=False)
Xb_test = Xb_test.reindex(columns=Xb_train.columns, fill_value=0)

print("Train size:", len(idx_train), "| Test size:", len(idx_test))
print("Fraud rate — train:", round(y_train.mean(), 3), "| test:", round(y_test.mean(), 3))
print("Encoded train/test shapes:", Xb_train.shape, Xb_test.shape)
```

## 9. Model Comparison

The primary comparison uses only the conservative feature whitelist. Graph-derived features, embeddings, shared-identifier counts, chargeback rate and refund ratio are excluded because their availability and label-independent provenance are not established.

We compare Logistic Regression, Random Forest and HistGradientBoosting on the same stratified holdout. This is a synthetic-data experiment, not evidence of production performance. For deployment-oriented evaluation, use a time-based split and a validation set that reflects the intended prediction time.


```python
def evaluate(name, model, X_train, X_test, y_train, y_test, scale=False):
    """Fit on training data and evaluate once on the held-out test data."""
    if scale:
        scaler = StandardScaler()
        X_train_fit = scaler.fit_transform(X_train)
        X_test_fit = scaler.transform(X_test)
    else:
        X_train_fit, X_test_fit = X_train, X_test

    model.fit(X_train_fit, y_train)
    proba = model.predict_proba(X_test_fit)[:, 1]
    pred = model.predict(X_test_fit)

    auc = roc_auc_score(y_test, proba)
    f1 = f1_score(y_test, pred, zero_division=0)
    print(f"{name:28s}  ROC-AUC = {auc:.4f}   F1 = {f1:.4f}")
    return {"name": name, "model": model, "auc": auc, "f1": f1,
            "proba": proba, "pred": pred}
```

```python
print("--- Leakage-reduced behavioral feature comparison ---")
results_behavioral = [
    evaluate("Logistic Regression", LogisticRegression(max_iter=500, class_weight="balanced"),
             Xb_train, Xb_test, y_train, y_test, scale=True),
    evaluate("Random Forest", RandomForestClassifier(
        n_estimators=200, max_depth=12, class_weight="balanced",
        random_state=42, n_jobs=-1), Xb_train, Xb_test, y_train, y_test),
    evaluate("HistGradientBoosting", HistGradientBoostingClassifier(
        max_iter=200, random_state=42, class_weight="balanced"),
             Xb_train, Xb_test, y_train, y_test),
]
```

```python
# Deliberately no "Full (+ graph)" model is trained here:
# its input graph is label-derived, so such scores would be misleading.
print("Graph and embedding features are excluded from predictive model training.")
```

```python
comparison = pd.DataFrame([
    {"Feature set": "Leakage-reduced behavioral", **{k: r[k] for k in ("name", "auc", "f1")}}
    for r in results_behavioral
]).rename(columns={"name": "Model", "auc": "ROC-AUC", "f1": "F1-score"})

comparison.sort_values("ROC-AUC", ascending=False)
```

**Interpretation:** Only the leakage-reduced behavioral feature set is evaluated as a predictive model. The graph-based features and embeddings are excluded because the supplied edge list is label-derived. Results should be reported as performance on this synthetic holdout, not as guaranteed real-world performance.


## 10. Final Model Selection & Evaluation

For a simple baseline workflow, the Random Forest is displayed below. For a rigorous final model choice, compare models and select hyperparameters using a separate validation set or cross-validation, then evaluate the selected model once on the untouched test set.


```python
final = next(r for r in results_behavioral if r["name"] == "Random Forest")
print(f"Final model: {final['name']}")
print(f"ROC-AUC: {final['auc']:.4f}")
print(f"F1-score: {final['f1']:.4f}")

print("\nFull classification report:")
print(classification_report(y_test, final["pred"], target_names=["Legit", "Fraud"]))
```

## 11. Conclusion

- The dataset is synthetic; conclusions describe patterns in this generated dataset and should not be generalized to real merchants without external validation.
- Graph edges were generated using fraud labels. Therefore, graph connectivity, edge-derived features and graph embeddings are excluded from the primary predictive model.
- Shared-identifier counts, chargeback rate and refund ratio are also excluded until their generation process and availability at prediction time are independently verified.
- The primary model uses a conservative set of merchant profile and transaction-volume/value features, with categorical encoding performed after the train/test split.
- The reported metrics must be regenerated by running the notebook from a clean kernel. They represent a stratified holdout result, not a production guarantee.
- Before deployment-oriented claims, validate with a time-based split, a documented prediction cutoff, leakage-safe feature generation, and external or independently generated data.

**Limitations:** Synthetic feature distributions may themselves be generated conditionally on `is_fraud`. Confirm the data-generation process for every retained predictor. If a retained feature was calculated using the label or future information, remove it or regenerate it from pre-prediction data.
